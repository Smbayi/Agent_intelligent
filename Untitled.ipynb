# Agent intelligent

```python
import pygame
import random
import time
import math

# Initialisation
pygame.init()

# Dimensions
WIDTH, HEIGHT = 800, 400
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Agent Aspirateur Intelligent - Version Finale")

# Couleurs
WHITE = (245, 245, 245)
BLACK = (0, 0, 0)
BROWN = (139, 69, 19)
GREY = (200, 200, 200)
BLUE = (100, 149, 237)
RED = (255, 0, 0)
DARK_GREY = (169, 169, 169)

# Police
font = pygame.font.SysFont("arial", 24)

# Zones : Chambre A | Couloir | Chambre B
room_A = pygame.Rect(0, 0, WIDTH // 3, HEIGHT)
hallway = pygame.Rect(WIDTH // 3, 0, WIDTH // 3, HEIGHT)
room_B = pygame.Rect(2 * WIDTH // 3, 0, WIDTH // 3, HEIGHT)

# Position initiale de l’agent
vacuum_x = hallway.centerx
vacuum_y = hallway.centery
vacuum_speed = 4  # vitesse augmentée

# Génération des déchets (5 par chambre)
def generate_dirt(room):
    return [(random.randint(room.left + 50, room.right - 50),
             random.randint(80, HEIGHT - 80)) for _ in range(5)]

# États initiaux
dirt_A = generate_dirt(room_A)
dirt_B = []
current_room = "A"
cleaning = False
pause_in_hallway = False
last_dirt_spawn = time.time()
cycle_start = time.time()

clock = pygame.time.Clock()
running = True

# === Fonctions de dessin ===
def draw_environment():
    screen.fill(WHITE)
    pygame.draw.rect(screen, BLUE, room_A)
    pygame.draw.rect(screen, DARK_GREY, hallway)
    pygame.draw.rect(screen, BLUE, room_B)
    pygame.draw.rect(screen, BLACK, room_A, 3)
    pygame.draw.rect(screen, BLACK, hallway, 3)
    pygame.draw.rect(screen, BLACK, room_B, 3)
    screen.blit(font.render("Chambre A", True, BLACK), (room_A.centerx - 60, 20))
    screen.blit(font.render("entrepot", True, BLACK), (hallway.centerx - 40, 20))
    screen.blit(font.render("Chambre B", True, BLACK), (room_B.centerx - 60, 20))

def draw_dirt(dirt_list):
    for (x, y) in dirt_list:
        pygame.draw.circle(screen, BROWN, (x, y), 8)

def draw_vacuum(x, y):
    pygame.draw.rect(screen, RED, (x - 20, y - 10, 40, 20))
    pygame.draw.circle(screen, BLACK, (x - 15, y + 10), 6)
    pygame.draw.circle(screen, BLACK, (x + 15, y + 10), 6)

def move_towards(target_x, target_y, speed):
    global vacuum_x, vacuum_y
    dx = target_x - vacuum_x
    dy = target_y - vacuum_y
    distance = math.hypot(dx, dy)
    if distance > speed:
        vacuum_x += speed * dx / distance
        vacuum_y += speed * dy / distance
        return False
    else:
        vacuum_x, vacuum_y = target_x, target_y
        return True

# === Boucle principale ===
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    draw_environment()

    # Apparition de nouveaux déchets toutes les 5 secondes dans UNE seule chambre
    if time.time() - last_dirt_spawn >= 5 and not cleaning and not pause_in_hallway:
        if random.choice(["A", "B"]) == "A":
            dirt_A = generate_dirt(room_A)
            dirt_B = []
        else:
            dirt_B = generate_dirt(room_B)
            dirt_A = []
        last_dirt_spawn = time.time()

    # Si les deux chambres sont propres → aller au couloir et attendre
    if not dirt_A and not dirt_B and not pause_in_hallway:
        current_room = None
        pause_in_hallway = True
        pause_start = time.time()

    if pause_in_hallway:
        move_towards(hallway.centerx, hallway.centery, vacuum_speed)
        if time.time() - pause_start >= 5:
            pause_in_hallway = False

    # Nettoyage chambre A
    if dirt_A and not pause_in_hallway:
        current_room = "A"
        if dirt_A:
            target = dirt_A[0]
            if move_towards(target[0], target[1], vacuum_speed):
                dirt_A.pop(0)
                time.sleep(0.2)  # nettoyage rapide
        if not dirt_A:
            current_room = "B"

    # Nettoyage chambre B
    elif dirt_B and not pause_in_hallway:
        current_room = "B"
        if dirt_B:
            target = dirt_B[0]
            if move_towards(target[0], target[1], vacuum_speed):
                dirt_B.pop(0)
                time.sleep(0.2)
        if not dirt_B:
            current_room = None

    # Dessin
    draw_dirt(dirt_A)
    draw_dirt(dirt_B)
    draw_vacuum(int(vacuum_x), int(vacuum_y))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```
